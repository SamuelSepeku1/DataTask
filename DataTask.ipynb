Relevant Information:
 -- Description 
 This data set consists of three types of entities:
  (a) the specification of an auto in terms of various characteristics 

  (b) its assigned insurance risk rating, 
  
  (c) its normalized losses in use as compared to other cars. The second rating corresponds to the degree to which the auto is more risky than its price indicates. Cars are initially assigned a risk factor symbol associated with its price. Then, if it is more risky (or less), this symbol is adjusted by moving it up (or down) the scale. Actuarians call this process "symboling". A value of +3 indicates that the auto is risky, -3 that it is probably pretty safe.

#	Column	Meaning

DATA PREPROCESSING

This involves structuring data into an approptiate format (data type) removing duplicates and filling null values. 


```python
# replace the value "?" with "" across all columns for proper data formatting
data=data.replace('?','')

target_columns=['normalized-losses','bore','stroke','horsepower','peak-rpm','price']
for x in target_columns:
    data[x]=pd.to_numeric(data[x])
    print(f"data type of {x} is:",data[x].dtype)

#rechecking data types for all columns in the dataset
print("\n\n",data.dtypes)


```

```python
#checking for null values after replacement and changing data types
data.isna().sum()
```

```python
#filling null values in the dataset
a=data['normalized-losses'].median()
data['normalized-losses']=data['normalized-losses'].fillna(a,inplace=True)

b=data['bore'].median()
data['bore']=data['bore'].fillna(b,inplace=True)

c=data['stroke'].median()
data['stroke']=data['stroke'].fillna(c,inplace=True)

x=data['horsepower'].median()
data['horsepower']=data['horsepower'].fillna(x,inplace=True)

y=data['peak-rpm'].median()
data['peak-rpm']=data['peak-rpm'].fillna(y,inplace=True)

z=data['price'].median()
data['price']=data['price'].fillna(z,inplace=True)
```

```python
#checking for null values after imputation
data.isna().sum()
```

```python
data.drop_duplicates()
```

IDENTIFYING THE CORRELATION OR RELATIONSHPS BETWEEN VARIABLES IN THE DATASET

```python
x=['symboling', 'normalized-losses', 'make', 'fuel-type', 'aspiration',
       'num-of-doors', 'body-style', 'drive-wheels', 'engine-location',
       'wheel-base', 'length', 'width', 'height', 'curb-weight', 'engine-type',
       'num-of-cylinders', 'engine-size', 'fuel-system', 'bore', 'stroke',
       'compression-ratio', 'horsepower', 'peak-rpm', 'city-mpg',
       'highway-mpg', 'price']

data_cols=[]

for col in x:
    if (data[col].dtype=='float64' or data[col].dtype=='int64'):
        data_cols.append(col)


print(data_cols)

sns.heatmap(data[data_cols].corr())
```
